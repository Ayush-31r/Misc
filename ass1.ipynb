```python
import numpy as np
import matplotlib.pyplot as plt

Gt = 1
Gr = 1
L = 1
fc = 900e6
c = 3e8
lambda_c = c / fc

d_km = np.arange(0.1, 20.1, 0.1)
d_m = d_km * 1000

fspl_linear = (Gt * Gr * lambda_c**2) / ((4 * np.pi * d_m)**2 * L)
fspl_dB = 10 * np.log10(fspl_linear)

plt.figure(figsize=(10, 6))
plt.plot(d_km, -fspl_dB, 'b-', linewidth=2.5)
plt.xlabel('Distance (km)', fontsize=13, fontweight='bold')
plt.ylabel('Path Loss (dB)', fontsize=13, fontweight='bold')
plt.title('Free Space Propagation Model', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.4, linestyle='--')
plt.xlim([0, 20])
plt.tight_layout()
plt.savefig('freespace_model.png', dpi=300, bbox_inches='tight')
plt.show()

print("=" * 60)
print(" FREE SPACE PROPAGATION MODEL - ANALYSIS RESULTS ")
print("=" * 60)
print(f"\nSystem Parameters:")
print(f" Carrier Frequency: {fc / 1e6:.0f} MHz")
print(f" Wavelength: {lambda_c:.4f} m")
print(f" Transmit Antenna Gain (Gt): {Gt} (0 dBi)")
print(f" Receive Antenna Gain (Gr): {Gr} (0 dBi)")
print(f" System Loss (L): {L} (0 dB)")

print("\n" + "=" * 60)
print(" PATH LOSS AT KEY DISTANCES ")
print("=" * 60)

distances = [1, 5, 10, 15, 20]
for dist in distances:
    idx = np.where(np.isclose(d_km, dist))[0][0]
    print(f" At d = {dist:2d} km: Path Loss = {-fspl_dB[idx]:8.2f} dB")

print("\n" + "=" * 60)
```
